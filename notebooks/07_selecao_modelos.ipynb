```python
results = json.loads((OUTPUT / "results.json").read_text())
display(results["metadata"])
rows = []
for name in ["logistic", "decision_tree"]:
    result = results[name]
    print(name, result["best_params"], "AP CV:", result["best_cv_ap"], "limiar:", result["threshold"])
    rows.append({"model": name, "best_cv_ap": result["best_cv_ap"],
                 "threshold": result["threshold"], "validation_best_f1": result["validation_best_f1"]})
display(pd.DataFrame(rows))
```
